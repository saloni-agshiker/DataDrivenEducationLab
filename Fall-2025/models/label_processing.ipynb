# Labeling Processing
This file contains calculation of average vote and majority vote of sentiment analysis manual labeling portion.

```python
import numpy as np
from statsmodels.stats.inter_rater import fleiss_kappa, aggregate_raters
import pandas as pd
import csv

df = pd.read_csv("labels_100.csv")
```

```python
clara = []
ethan = []
saloni = []
with open('labels_100.csv', newline='') as csvfile:
    spamreader = csv.DictReader(csvfile)
    for row in spamreader:
        clara.append(row['Clara'])
        ethan.append(row['Ethan'])
        saloni.append(row['Saloni'])
```

```python
from sklearn.metrics import cohen_kappa_score
```

```python
cohen_kappa_score(clara[:1000],ethan[:1000]) # calculate first 1000 labels
```

```python
cohen_kappa_score(clara[:1000],saloni[:1000])
```

```python
cohen_kappa_score(ethan[:1000],saloni[:1000])
```

```python
contingency_table, sum_ratings = aggregate_raters(np.transpose([clara[:1000],ethan[:1000],saloni[:1000]]))
fleiss_kappa(contingency_table, method='fleiss')
```

```python
c = [int(ci) for ci in clara[:1000]]
e = [int(ei) for ei in ethan[:1000]]
s = [int(si) for si in saloni[:1000]]
```

## Calculation for Average Vote and Majority Vote

```python
avg_labels = [(int(ci) + int(ei) + int(si)) / 3 for ci, ei, si in zip(c, e, s)]
avg_vote = [round(x) for x in avg_labels]
```

```python
from scipy import stats
stacked = np.stack([np.array(clara[:999]).astype(int),np.array(ethan[:999]).astype(int),np.array(saloni[:999]).astype(int)])
majority_vote = stats.mode(stacked, axis=0).mode
```
