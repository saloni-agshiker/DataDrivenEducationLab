```python
import json
import re
from textblob import *
import numpy as np
```

# Data Cleaning

```python
#Cleaning html tags from posts
def cleanhtml(raw):
    cleanr = re.compile('<.*?>')
    cleantext = re.sub(cleanr, '', raw)
    return cleantext
```

```python
#Cleaning pin tags from posts
def cleanpin(raw):
    cleantext = re.sub('#pin','',raw)
    return cleantext
```

```python
def merge_content(out_dict):
    new_list = []
    for i in list(out_dict.values()):
        new_list.extend(i)
    return new_list
```

```python
#Cleaning raw data and outputing post contents to a json file
raw = json.load(open('ISYE6105_Spring_2019.json','r'))
post_list = raw[0]['posts']
output_dict = {}
time_dict = {}
import csv
f = open('time_sentiment.csv', 'w')
writer = csv.writer(f)
writer.writerow(['time', 'sentiment'])
for i in post_list:
    folders = i['folders']
    latest = i['history'][-1]
    raw_content = latest['content']
    time = latest['created'][0:7]
    cleaned_content = cleanpin(cleanhtml(raw_content.replace('\u00a0',' ').replace('\n',' '))).strip()
    x = TextBlob(cleaned_content)
    sent = x.sentiment.polarity
    writer.writerow([time, sent])
    for j in folders:
        if j not in output_dict:
            output_dict[j] = [cleaned_content]
        else:
            output_dict[j].append(cleaned_content)
json.dump(output_dict,open('post_content.json','w'))
f.close()
```

## Simple NLP analysis on cleaned post content

```python
avg_sentiment = {}
individual_sentiment = {}
neg_sentiment = {}
for i in output_dict:
    avg_sentiment[i] = 0
    individual_sentiment[i] = []
    content_list = output_dict[i]
    neg_sentiment[i] = []
    count = 0
    for j in content_list:
        x = TextBlob(j)
        avg_sentiment[i] += x.sentiment.polarity
        count += 1
        individual_sentiment[i].append(x.sentiment.polarity)
        if x.sentiment.polarity < 0:
            neg_sentiment[i].append(j)
    avg_sentiment[i] = avg_sentiment[i]/count
post_count = {i:len(individual_sentiment[i]) for i in individual_sentiment}
neg_post_count = {i:len(neg_sentiment[i]) for i in neg_sentiment}
std_dev_sentiment = {}
for i in individual_sentiment:
    nums = individual_sentiment[i]
    std_dev_sentiment[i] = np.std(nums)
#outputing negative post contents to a json file
#json.dump(neg_sentiment,open('neg_post_content.json','w'))
```

## Tokenization and Stemming

```python
import nltk
from nltk.stem.snowball import SnowballStemmer

stopwords = nltk.corpus.stopwords.words('english')
stemmer = SnowballStemmer("english")

# tokenization without stemming
def tokenization(text):
    tokens = [word.lower() for sent in nltk.sent_tokenize(text) for word in nltk.word_tokenize(sent) if word not in stopwords]
    filtered_tokens = []
    for token in tokens:
        if re.search('[a-zA-Z]', token):
            result = re.search('.*([a-zA-Z]).*', token).group()
            filtered_tokens.append(result)
    return filtered_tokens
# tokenization with stemming
def tokenization_and_stemming(text):
    tokens = [word.lower() for sent in nltk.sent_tokenize(text) for word in nltk.word_tokenize(sent) if word not in stopwords]
    filtered_tokens = []
    for token in tokens:
        if re.search('[a-zA-Z]', token):
            result = re.search('.*([a-zA-Z]).*', token).group()
            filtered_tokens.append(result)
    #stemming
    stems = [stemmer.stem(t) for t in filtered_tokens]
    return stems
```

```python
posts_tokenized = []
posts_tokenized_stemmed = []
posts = merge_content(output_dict)
for post in posts:
    #with stemming
    tokenized_and_stemmed_results = tokenization_and_stemming(post)
    posts_tokenized_stemmed.extend(tokenized_and_stemmed_results)
    #without stemming
    tokenized_results = tokenization(post)
    posts_tokenized.extend(tokenized_results)
```

```python
# create a mapping from stemmed words to original words for interpretation
vocab_frame_dict = {posts_tokenized_stemmed[x]:posts_tokenized[x] for x in range(len(posts_tokenized_stemmed))}
vocab_frame_dict
```

```python
# word frequency
from nltk.probability import FreqDist
fdist1 = FreqDist(posts_tokenized_stemmed)
fdist2 = FreqDist(posts_tokenized)
fdist1_pop = fdist1.most_common(10)
fdist2_pop = fdist2.most_common(10)
print(fdist1_pop)
print(fdist2_pop)
```
