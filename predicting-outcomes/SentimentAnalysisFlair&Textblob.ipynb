```python
import io
import pandas as pd
from textblob import TextBlob
import csv

# download flair library #
import torch
import flair
from flair.models import TextClassifier
from flair.data import Sentence
```

```python
data = pd.read_csv("/Users/malavpatel/Documents/GA_Tech/VIP_Data_Driven/piazza data anonymized.csv", usecols = ['Submission HTML Removed_coding','Created_At'])
text = data.values.tolist()
```

```python
#file import

f = open('/Users/malavpatel/Documents/GA_Tech/VIP_Data_Driven/Flair_Textblob.csv', 'w')
writer = csv.writer(f)
```

```python
#Generating a list of week numbers

data['Created_At'] = pd.to_datetime(data.Created_At)
data['Created_At'] = pd.to_datetime(data.Created_At).dt.tz_localize(None)
list = data.Created_At.dt.isocalendar().week
```

```python
for i in range(len(text)): 
    Matrix = []
    classifier = TextClassifier.load('en-sentiment')
    sentence = Sentence(text[i][1])
    classifier.predict(sentence)
    blob = TextBlob(text[i][1])
    writer.writerow([list[i], text[i][0], text[i][1], sentence.labels, blob.sentiment.polarity, blob.sentiment.subjectivity])
```
