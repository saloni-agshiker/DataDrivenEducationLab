```python
# import pandas as pd
# from datetime import datetime, timedelta

import pandas as pd
from datetime import datetime, timedelta
import json

# Read CSV file into a DataFrame
df = pd.read_csv("piazzadataanonymized.csv")

# Convert "Created At" column to datetime format
df["Created At"] = pd.to_datetime(df["Created At"])

# Add new columns for month and week
df["Month"] = df["Created At"].dt.strftime('%Y-%m')
df["Week"] = df["Created At"].dt.strftime('%Y-%U')

# Initialize variables for averaging
total_time_diff = timedelta(0)
count = 0

# Group the DataFrame by "Post Number", month, and week
grouped = df.groupby(["Post Number", "Month", "Week"])

weekly_data = []
i = 0
# Iterate over each group
for (name, month, week), group in grouped:
    # Find the index of the first "Student" user_role post
    student_index = group[group["user_role"] == "Student"].index
    if len(student_index) == 0:
        continue
    student_index = student_index[0]

    # Find the index of the next "TA" user_role post with "reply_to_followup" part of post
    ta_index = group[(group["user_role"] == "TA") & (group["Part of Post"] == "reply_to_followup")].index
    if len(ta_index) == 0:
        continue
    ta_index = ta_index[0]

    # Calculate the time difference between the two posts
    time_diff = group.loc[ta_index, "Created At"] - group.loc[student_index, "Created At"]
    
    # Check if time difference is negative, and if so, skip this post
    if time_diff < timedelta(0):
        continue
    
    # Add to variables for averaging
    total_time_diff += time_diff
    count += 1
    
    # If this is the last post for the week, print the average time difference
    if ta_index == group.index[-1]:
        i += 1
        String = "Week " + str (i)
        average_time_diff = total_time_diff / count
        average_time_diff_hours = int(average_time_diff.total_seconds() / 3600)
        print(f"Average time difference for week {week} in {month}: {average_time_diff_hours} hours")
        myDict = {}
        myDict["Week"] = String
        myDict["Time"] = average_time_diff_hours
        weekly_data.append(myDict)
# Create a dictionary with the "data" key and the weekly_data list as its value
data_dict = {"data": weekly_data}

# Convert the dictionary to a JSON-formatted string
json_str = json.dumps(data_dict)

with open("weekly_data.json", "w") as f:
    f.write(json_str)

```

```python
# read the CSV file and extract the "Created At" column
df1 = pd.read_csv('piazzadataanonymized.csv')
created_at = pd.to_datetime(df1['Created At'])

# drop duplicates based on the "Post Number" column
df_unique = df1.drop_duplicates(subset=['Post Number'])

# group the unique posts by month
posts_by_month = df_unique.groupby(created_at.dt.to_period('M'))

# iterate over the groups and print the message for each post
for month, posts in posts_by_month:
    for post in posts['Post Number']:
        print(f"This post {post} was created in {month}")
```
