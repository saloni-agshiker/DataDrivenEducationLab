```python
participantStatus = pd.read_csv("https://gist.githubusercontent.com/mikster36/bf6b3265b74f67b219f1aeccc5da683b/raw/2e6dbd8072330b92ba0d04ff74b238434ec732ad/Participant_Status.csv")
geneticStatus = pd.read_csv("https://gist.githubusercontent.com/rezareimoo/3ffc36a4a122c2cf6a1076df13facc25/raw/9b3d22f24ab66e87c99f7a3fe7289790ddf57ab5/genetic_status.csv")
participantStatus = participantStatus[["PATNO","COHORT"]]
merged_df = pd.merge(participantStatus, geneticStatus, on='PATNO')
merged_df = merged_df.drop('PATNO', axis=1)
merged_df = merged_df[merged_df['COHORT'] <= 2]
#merged_df = merged_df[~merged_df.apply(lambda x: x.astype(str).str.contains('3')).any(axis=1)]
#merged_df = merged_df[~merged_df.apply(lambda x: x.astype(str).str.contains('4')).any(axis=1)]
merged_df = merged_df.dropna()
```
