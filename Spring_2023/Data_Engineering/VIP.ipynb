```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
uploaded = files.upload()
# importing libraries and the csv file
```

```python
# Load the CSV file into a pandas DataFrame
df = pd.read_csv('MetacognitivePresenceCodedData.csv')

# Extract the "Sum of MGP" column
sum_of_mgp = df['Sum of MGP']

# Use the value_counts() method to count the frequency of each value
counts = sum_of_mgp.value_counts()

# Create a bar chart of the counts
plt.bar(counts.index, counts.values)

# Set the chart title and axis labels
plt.title('Sum of MGP Value Counts')
plt.xlabel('Value')
plt.ylabel('Count')

# Show the chart
plt.show()
```

```python
# Extract the "Knowledge of Cognition (KC)" column
kc = df['Knowledge of Cognition (KC)']

# Use the value_counts() method to count the frequency of each value
counts = kc.value_counts()

# Create a bar chart of the counts
plt.bar(counts.index, counts.values)
colors = ['blue', 'orange', 'green', 'red', 'purple', 'brown']

# Create a bar chart of the counts with custom colors
plt.bar(counts.index, counts.values, color=colors[:len(counts)])

# Set the chart title and axis labels
plt.title('Knowledge of Cognition Value Counts')
plt.xlabel('Value')
plt.ylabel('Count')

# Add the count value at the top of each bar
for i, count in enumerate(counts):
    plt.text(counts.index[i], count + 0.5, str(count), ha='center')
# Show the chart
plt.show()
```

```python
# Create a new column "Regulation of Cognition" by summing the RC columns
df = pd.read_csv('MetacognitivePresenceCodedData.csv')
rc_cols = ['Planning (RC)', 'Information/Process management (RC)', 
           'Monitoring (RC)', 'Debugging (RC)', 'Evaluation (RC)']
df['Regulation of Cognition'] = df[rc_cols].sum(axis=1)

# Use the value_counts() method to count the frequency of each value
counts = df['Regulation of Cognition'].value_counts()


# Create a bar chart of the counts
plt.bar(counts.index, counts.values)

colors = ['blue', 'orange', 'green', 'red', 'purple', 'brown']

# Create a bar chart of the counts with custom colors
plt.bar(counts.index, counts.values, color=colors[:len(counts)])

# Set the chart title and axis labels
plt.title('Regulation of Cognition Value Counts')
plt.xlabel('Value')
plt.ylabel('Count')

# Add the count value at the top of each bar
for i, count in enumerate(counts):
    plt.text(counts.index[i], count + 0.5, str(count), ha='center')

# Show the chart
plt.show()
```
