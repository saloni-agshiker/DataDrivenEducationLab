```python
import pandas as pd
import pyLDAvis.gensim_models as gensimvis
import pyLDAvis
from gensim import corpora
from gensim.models import LdaModel
import nltk
from nltk.corpus import stopwords
import string
```

```python
df = pd.read_csv('CS1301_filtered.csv')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words("english"))
texts = df["cleaned_text"].dropna().astype(str).apply(lambda x: x.split())
processed_texts = [
    [word for word in doc if word.lower() not in stop_words and word not in string.punctuation]
    for doc in texts
]
```

```python
dictionary = corpora.Dictionary(processed_texts)
corpus = [dictionary.doc2bow(text) for text in processed_texts]
```

```python
lda_model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=5, passes=10, random_state=42)
```

```python
pyLDAvis.enable_notebook()  # or use pyLDAvis.show() in a script
vis = gensimvis.prepare(lda_model, corpus, dictionary)
vis
```

```python

# Display top terms with weights for each topic
topn = 10
for topic_id in range(lda_model.num_topics):
    print(f"Topic #{topic_id}:")
    terms = lda_model.show_topic(topic_id, topn=topn)
    for term, weight in terms:
        print(f"  {term:<15} {weight:.4f}")
    print()
```
