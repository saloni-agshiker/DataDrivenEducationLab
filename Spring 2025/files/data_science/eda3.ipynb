```python
def filter_by_course(course_name):

    # Filter rows where course_id contains any of the keywords
    filtered_df = df[df["course_id"].astype(str).str.contains(course_name, na=False)]

    # Save to a new CSV file
    filtered_df.to_csv(f'{course_name}_filtered.csv', index=False)
```

```python
filter_by_course('CS1301')
filter_by_course('ISYE6501')
filter_by_course('CSE6040')
filter_by_course('MGT6203')
filter_by_course('HDM1')
```
